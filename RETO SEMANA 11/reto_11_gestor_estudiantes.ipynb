````python
def info_general(df_estudiantes: pd.DataFrame, df_calificaciones: pd.DataFrame) -> Dict:
    """
    Genera información general del sistema.
    """
    resultado = {
        "total_estudiantes": int(df_estudiantes["boleta"].nunique()),
        "total_registros_calif": int(len(df_calificaciones)),
        "semestres": sorted(df_estudiantes["semestre"].unique().tolist()),
        "materias_con_registros": int(df_calificaciones["materia_id"].nunique())
    }
    return resultado
````

````python
def validar_datos(df_calificaciones: pd.DataFrame) -> Dict:
    """
    Valida la integridad de los datos.
    """
    cols_calif = ["parcial_1", "parcial_2", "final"]

    # Filas que tienen al menos un nulo en columnas de calificación
    registros_con_nulos = int(df_calificaciones[cols_calif].isna().any(axis=1).sum())

    # Máscara booleana: cualquier calificación fuera de [0, 10] (ignorando NaN)
    mask_fuera = (
        df_calificaciones[cols_calif]
        .apply(lambda s: (s < 0) | (s > 10))
        .any(axis=1)
    )
    calificaciones_fuera_rango = int(mask_fuera.sum())

    resultado = {
        "registros_con_nulos": registros_con_nulos,
        "calificaciones_fuera_rango": calificaciones_fuera_rango,
        "datos_validos": (registros_con_nulos == 0) and (calificaciones_fuera_rango == 0)
    }
    return resultado
````

````python
# PARTE 2: Consultas y Filtros

def buscar_estudiante(df_estudiantes: pd.DataFrame, criterio: str, valor: str) -> pd.DataFrame:
    """
    Busca estudiantes por diferentes criterios.
    """
    if criterio == "nombre":
        return df_estudiantes[df_estudiantes["nombre"].str.contains(valor, case=False, na=False)]
    elif criterio == "boleta":
        return df_estudiantes[df_estudiantes["boleta"] == valor]
    elif criterio == "semestre":
        return df_estudiantes[df_estudiantes["semestre"] == int(valor)]
    else:
        return df_estudiantes.iloc[0:0]  # DataFrame vacío con mismas columnas
````

````python
def obtener_kardex(boleta: str, df_estudiantes: pd.DataFrame,
                   df_calificaciones: pd.DataFrame, df_materias: pd.DataFrame) -> Dict:
    """
    Obtiene el kardex completo de un estudiante.
    """
    resultado = {
        "estudiante": None,
        "materias": None,
        "promedio_general": None,
        "creditos_cursados": None,
        "materias_aprobadas": None,
        "materias_reprobadas": None
    }

    # Datos del estudiante
    fila_est = df_estudiantes[df_estudiantes["boleta"] == boleta]
    if fila_est.empty:
        return resultado
    resultado["estudiante"] = fila_est.iloc[0].to_dict()

    # Calificaciones del estudiante, unidas con nombres y créditos
    califs = df_calificaciones[df_calificaciones["boleta"] == boleta].merge(
        df_materias[["materia_id", "nombre", "creditos"]],
        on="materia_id",
        how="left"
    ).copy()

    # Promedio por materia: media de los tres parciales (ignora NaN)
    califs["promedio_materia"] = califs[["parcial_1", "parcial_2", "final"]].mean(axis=1)

    # Columnas visibles en el kardex
    califs_display = califs[["nombre", "parcial_1", "parcial_2", "final", "promedio_materia", "creditos"]].copy()
    califs_display.columns = ["Materia", "Parcial 1", "Parcial 2", "Final", "Promedio", "Créditos"]

    resultado["materias"] = califs_display
    resultado["promedio_general"] = round(float(califs["promedio_materia"].mean()), 2)
    resultado["creditos_cursados"] = int(califs["creditos"].sum())
    resultado["materias_aprobadas"] = int((califs["promedio_materia"] >= 6.0).sum())
    resultado["materias_reprobadas"] = int((califs["promedio_materia"] < 6.0).sum())

    return resultado
````

````python
def filtrar_por_rendimiento(df_calificaciones: pd.DataFrame,
                            df_estudiantes: pd.DataFrame,
                            min_promedio: float = None,
                            max_promedio: float = None) -> pd.DataFrame:
    """
    Filtra estudiantes por rango de promedio.
    """
    # Promedio por materia → promedio general por boleta
    prom = (
        df_calificaciones
        .assign(promedio_materia=df_calificaciones[["parcial_1", "parcial_2", "final"]].mean(axis=1))
        .groupby("boleta", as_index=False)["promedio_materia"]
        .mean()
        .rename(columns={"promedio_materia": "promedio_general"})
    )

    # Aplicar filtros booleanos vectorizados
    mask = pd.Series([True] * len(prom))
    if min_promedio is not None:
        mask &= prom["promedio_general"] >= min_promedio
    if max_promedio is not None:
        mask &= prom["promedio_general"] <= max_promedio

    prom_filtrado = prom[mask]

    return prom_filtrado.merge(df_estudiantes, on="boleta", how="left")
````

````python
# PARTE 3: Cálculos y Estadísticas

def calcular_promedio_materia(df_calificaciones: pd.DataFrame, materia_id: str) -> Dict:
    """
    Calcula estadísticas de una materia.
    """
    df_mat = df_calificaciones[df_calificaciones["materia_id"] == materia_id].copy()
    df_mat["promedio_materia"] = df_mat[["parcial_1", "parcial_2", "final"]].mean(axis=1)

    tasa = float((df_mat["promedio_materia"] >= 6.0).sum()) / max(len(df_mat), 1) * 100

    resultado = {
        "materia": materia_id,
        "inscritos": int(len(df_mat)),
        "promedio_parcial1": round(float(df_mat["parcial_1"].mean()), 2),
        "promedio_parcial2": round(float(df_mat["parcial_2"].mean()), 2),
        "promedio_final": round(float(df_mat["final"].mean()), 2),
        "promedio_general": round(float(df_mat["promedio_materia"].mean()), 2),
        "tasa_aprobacion": round(tasa, 1),
        "calificacion_maxima": round(float(df_mat["promedio_materia"].max()), 2),
        "calificacion_minima": round(float(df_mat["promedio_materia"].min()), 2)
    }
    return resultado
````

````python
def ranking_estudiantes(df_calificaciones: pd.DataFrame,
                        df_estudiantes: pd.DataFrame,
                        top_n: int = 10) -> pd.DataFrame:
    """
    Genera ranking de mejores estudiantes por promedio.
    """
    ranking = (
        df_calificaciones
        .assign(promedio_materia=df_calificaciones[["parcial_1", "parcial_2", "final"]].mean(axis=1))
        .groupby("boleta", as_index=False)["promedio_materia"]
        .mean()
        .rename(columns={"promedio_materia": "promedio_general"})
        .sort_values("promedio_general", ascending=False)
        .head(top_n)
        .merge(df_estudiantes[["boleta", "nombre", "semestre"]], on="boleta", how="left")
        .reset_index(drop=True)
    )
    ranking.insert(0, "posicion", ranking.index + 1)
    ranking["promedio_general"] = ranking["promedio_general"].round(2)
    return ranking
````

````python
def estadisticas_por_semestre(df_estudiantes: pd.DataFrame,
                              df_calificaciones: pd.DataFrame) -> pd.DataFrame:
    """
    Calcula estadísticas agrupadas por semestre.
    """
    # Promedio general por estudiante
    prom_est = (
        df_calificaciones
        .assign(prom_mat=df_calificaciones[["parcial_1", "parcial_2", "final"]].mean(axis=1))
        .groupby("boleta", as_index=False)["prom_mat"]
        .mean()
        .rename(columns={"prom_mat": "promedio_general"})
    )

    # Unir con semestre
    df_merged = prom_est.merge(df_estudiantes[["boleta", "semestre"]], on="boleta", how="left")

    # Columna auxiliar: aprobado si promedio >= 6
    df_merged["aprobado"] = df_merged["promedio_general"] >= 6.0

    stats = (
        df_merged.groupby("semestre")
        .agg(
            estudiantes=("boleta", "count"),
            promedio=("promedio_general", "mean"),
            tasa_aprobacion=("aprobado", "mean"),
            mejor_promedio=("promedio_general", "max"),
            peor_promedio=("promedio_general", "min")
        )
        .round({"promedio": 2, "tasa_aprobacion": 3, "mejor_promedio": 2, "peor_promedio": 2})
    )
    stats["tasa_aprobacion"] = (stats["tasa_aprobacion"] * 100).round(1).astype(str) + "%"
    return stats
````

````python
# PARTE 4: Identificación de Riesgo y Reportes

def identificar_estudiantes_riesgo(df_calificaciones: pd.DataFrame,
                                   df_estudiantes: pd.DataFrame,
                                   umbral_promedio: float = 7.0,
                                   max_reprobadas: int = 2) -> pd.DataFrame:
    """
    Identifica estudiantes en riesgo académico.
    """
    df_c = df_calificaciones.assign(
        prom_mat=df_calificaciones[["parcial_1", "parcial_2", "final"]].mean(axis=1)
    )

    # Promedio general y conteo de reprobadas por boleta en un solo groupby
    resumen = df_c.groupby("boleta").agg(
        promedio=("prom_mat", "mean"),
        reprobadas=("prom_mat", lambda x: (x < 6.0).sum())
    ).reset_index()

    # Máscaras booleanas vectorizadas
    bajo_prom = resumen["promedio"] < umbral_promedio
    muchas_rep = resumen["reprobadas"] > max_reprobadas

    resumen_riesgo = resumen[bajo_prom | muchas_rep].copy()

    # Motivo condicional vectorizado con np.select
    condiciones = [bajo_prom[bajo_prom.index.isin(resumen_riesgo.index)] & muchas_rep[muchas_rep.index.isin(resumen_riesgo.index)],
                   bajo_prom[bajo_prom.index.isin(resumen_riesgo.index)],
                   muchas_rep[muchas_rep.index.isin(resumen_riesgo.index)]]

    # Re-calcular máscaras sobre el subconjunto ya filtrado
    resumen_riesgo["motivo"] = np.select(
        [
            (resumen_riesgo["promedio"] < umbral_promedio) & (resumen_riesgo["reprobadas"] > max_reprobadas),
            resumen_riesgo["promedio"] < umbral_promedio,
            resumen_riesgo["reprobadas"] > max_reprobadas
        ],
        ["Ambos", "Bajo promedio", "Mat. reprob."],
        default="Bajo promedio"
    )

    resumen_riesgo["promedio"] = resumen_riesgo["promedio"].round(2)

    return resumen_riesgo.merge(
        df_estudiantes[["boleta", "nombre"]], on="boleta", how="left"
    )[["boleta", "nombre", "promedio", "reprobadas", "motivo"]]
````

````python
def generar_reporte_academico(df_estudiantes: pd.DataFrame,
                              df_calificaciones: pd.DataFrame,
                              df_materias: pd.DataFrame) -> Dict:
    """
    Genera reporte académico completo.
    """
    # Promedio de cada materia para reporte por materia
    df_c = df_calificaciones.assign(
        prom_mat=df_calificaciones[["parcial_1", "parcial_2", "final"]].mean(axis=1)
    )

    prom_global = float(df_c["prom_mat"].mean())
    tasa_global = float((df_c["prom_mat"] >= 6.0).mean()) * 100

    por_materia = (
        df_c.groupby("materia_id")
        .agg(
            inscritos=("boleta", "count"),
            promedio=("prom_mat", "mean"),
            tasa_aprobacion=("prom_mat", lambda x: round((x >= 6.0).mean() * 100, 1))
        )
        .round({"promedio": 2})
        .merge(df_materias[["materia_id", "nombre"]], on="materia_id", how="left")
        .set_index("materia_id")
    )

    reporte = {
        "resumen_general": {
            "total_estudiantes": int(len(df_estudiantes)),
            "promedio_global": round(prom_global, 2),
            "tasa_aprobacion": round(tasa_global, 1)
        },
        "por_semestre": estadisticas_por_semestre(df_estudiantes, df_calificaciones),
        "por_materia": por_materia,
        "mejores_estudiantes": ranking_estudiantes(df_calificaciones, df_estudiantes, top_n=10),
        "estudiantes_riesgo": identificar_estudiantes_riesgo(df_calificaciones, df_estudiantes),
        "fecha_generacion": datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    }
    return reporte
````

````python
def exportar_kardex(boleta: str, kardex: Dict, formato: str = 'csv') -> str:
    """
    Exporta el kardex de un estudiante a archivo.
    Nombres de salida: kardex_{boleta}.csv / kardex_{boleta}.json
    """
    nombre_archivo = f"kardex_{boleta}.{formato}"

    if kardex["materias"] is None:
        raise ValueError(f"El kardex de la boleta {boleta} no contiene datos.")

    # Construir DataFrame con llaves técnicas exactas requeridas por el validador
    califs_raw = df_calificaciones[df_calificaciones["boleta"] == boleta].merge(
        df_materias[["materia_id", "nombre", "creditos"]], on="materia_id", how="left"
    ).copy()
    califs_raw["promedio"] = califs_raw[["parcial_1", "parcial_2", "final"]].mean(axis=1).round(2)

    export_df = califs_raw[["materia_id", "nombre", "parcial_1", "parcial_2", "final", "promedio"]].copy()

    if formato == "csv":
        export_df.to_csv(nombre_archivo, index=False, encoding="utf-8")

    elif formato == "json":
        with open(nombre_archivo, "w", encoding="utf-8") as f:
            json.dump(export_df.to_dict(orient="records"), f, ensure_ascii=False, indent=4)

    else:
        raise ValueError(f"Formato no soportado: {formato}. Usa 'csv' o 'json'.")

    print(f"✅ Kardex exportado: {nombre_archivo}")
    return nombre_archivo
````

### 1. Predictor de riesgo

````python
def predecir_riesgo_proximo_semestre(df_calificaciones: pd.DataFrame,
                                     df_estudiantes: pd.DataFrame) -> pd.DataFrame:
    """
    Predice estudiantes que podrían estar en riesgo el próximo semestre
    basándose en tendencias de calificaciones.

    Criterios vectorizados:
    - Tendencia decreciente: parcial_2 < parcial_1
    - Calificación final menor que parcial_1
    """
    df_c = df_calificaciones.copy()

    # Máscara 1: tendencia decreciente entre parciales (ignorando NaN)
    tendencia_baja = df_c["parcial_2"] < df_c["parcial_1"]

    # Máscara 2: final menor que el primer parcial
    caida_final = df_c["final"] < df_c["parcial_1"]

    # Combinar: ambos criterios por materia
    df_c["en_riesgo"] = (tendencia_baja & caida_final).astype(int)

    # Agregar por estudiante: porcentaje de materias con señal de riesgo
    riesgo_est = (
        df_c.groupby("boleta", as_index=False)["en_riesgo"]
        .agg(materias_en_riesgo="sum", total_materias="count")
    )
    riesgo_est["pct_riesgo"] = (riesgo_est["materias_en_riesgo"] / riesgo_est["total_materias"] * 100).round(1)

    # Considerar en riesgo si >30% de sus materias muestran tendencia negativa
    resultado = (
        riesgo_est[riesgo_est["pct_riesgo"] > 30]
        .merge(df_estudiantes[["boleta", "nombre", "semestre"]], on="boleta", how="left")
        .sort_values("pct_riesgo", ascending=False)
        .reset_index(drop=True)
    )
    return resultado[["boleta", "nombre", "semestre", "materias_en_riesgo", "total_materias", "pct_riesgo"]]
````

### 2. Comparador de rendimiento

````python
def comparar_estudiantes(boleta1: str, boleta2: str,
                         df_calificaciones: pd.DataFrame,
                         df_estudiantes: pd.DataFrame,
                         df_materias: pd.DataFrame) -> Dict:
    """
    Compara el rendimiento de dos estudiantes.
    """
    def _resumen(boleta):
        kardex = obtener_kardex(boleta, df_estudiantes, df_calificaciones, df_materias)
        return kardex

    k1 = _resumen(boleta1)
    k2 = _resumen(boleta2)

    # Materias en común
    if k1["materias"] is not None and k2["materias"] is not None:
        ids1 = set(df_calificaciones[df_calificaciones["boleta"] == boleta1]["materia_id"])
        ids2 = set(df_calificaciones[df_calificaciones["boleta"] == boleta2]["materia_id"])
        comunes = ids1 & ids2

        prom_comunes = (
            df_calificaciones[df_calificaciones["materia_id"].isin(comunes)]
            .assign(prom=df_calificaciones[["parcial_1", "parcial_2", "final"]].mean(axis=1))
            .groupby(["boleta", "materia_id"])["prom"]
            .mean()
            .unstack(level=0)
            .rename(columns={boleta1: k1["estudiante"]["nombre"], boleta2: k2["estudiante"]["nombre"]})
            .merge(df_materias[["materia_id", "nombre"]].set_index("materia_id"),
                   left_index=True, right_index=True, how="left")
            .reset_index(drop=True)
        )
    else:
        prom_comunes = pd.DataFrame()

    return {
        "estudiante_1": k1["estudiante"],
        "estudiante_2": k2["estudiante"],
        "promedio_1": k1["promedio_general"],
        "promedio_2": k2["promedio_general"],
        "ganador": k1["estudiante"]["nombre"] if (k1["promedio_general"] or 0) >= (k2["promedio_general"] or 0) else k2["estudiante"]["nombre"],
        "materias_comunes": prom_comunes
    }
````
